````python
import pandas as pd

data = {
    "Product": ["Laptop", "Mobile", "Headphones", "Tablet", "Smartwatch",
                "Laptop", "Mobile", "Headphones", "Tablet", "Smartwatch",
                "Laptop", "Mobile"],

    "Category": ["Electronics", "Electronics", "Accessories", "Electronics", "Accessories",
                 "Electronics", "Electronics", "Accessories", "Electronics", "Accessories",
                 "Electronics", "Electronics"],

    "Month": ["Jan", "Jan", "Feb", "Feb", "Mar",
              "Mar", "Apr", "Apr", "May", "May",
              "Jun", "Jun"],

    "Sales": [85000, 62000, 28000, 45000, 22000,
              92000, 70000, 32000, 51000, 27000,
              98000, 76000],

    "Units_Sold": [17, 31, 40, 15, 22,
                   18, 35, 45, 17, 26,
                   20, 38],

    "Rating": [4.5, 4.3, 4.1, 4.4, 4.2,
               4.6, 4.4, 4.0, 4.5, 4.3,
               4.7, 4.5]
}

df = pd.DataFrame(data)

df
````

````python
import matplotlib.pyplot as plt
import pandas as pd
````

````python
plt.figure(figsize=(10,5))
plt.bar(df['Product'],df['Sales'],color='green')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Sales of Each Product')
plt.show()
````

````python
plt.figure(figsize=(10,5))
plt.plot(df['Month'],df['Sales'],"go--")
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Sales for Each Month')
plt.show()
````

````python
plt.figure(figsize=(10,5))
plt.barh(df['Product'],df['Units_Sold'],color='green')
plt.xlabel('Units Sold')
plt.ylabel('Producr')
plt.title('Number of Units Solds For Each Product')
plt.show()
````

````python
plt.figure(figsize=(10,5))
plt.hist(df['Rating'],bins=5,color='green',edgecolor='black')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.title('Distribution of Product Ratings')
plt.show()
````

````python
plt.figure(figsize=(10,5))
plt.scatter(df['Units_Sold'],df['Sales'],s=110,c=df['Rating'])
plt.colorbar()
plt.xlabel('Units Sold')
plt.ylabel('Sales')
plt.title('Relationship Between Units Sold and Sales')
plt.grid('True',alpha=0.3)
plt.show()
````

- Q6. Pie Chart
Create a pie chart showing the total sales contribution of each product.

- Create a 2×2 subplot containing the following four charts using the given dataset:
1. Bar chart → Product-wise Sales
2. Line chart → Monthly Sales
3. Scatter plot → Units Sold vs Sales
4. Histogram → Distribution of Ratings

````python
plt.figure(figsize=(10,8))
plt.pie(df['Sales'],labels=df['Product'],autopct='%1.1f%%')
plt.title('Total Sales Contribution of Each Product')
plt.show()
````

````python
fig, axes = plt.subplots(2, 2, figsize=(10,6))

fig.suptitle("Sales Data", fontsize=24, fontweight = "bold")

axes[0,0].bar(df['Product'], df['Sales'],color='green')
axes[0,0].set_xlabel('Product')
axes[0,0].set_ylabel('Sales')
axes[0,0].set_title('Product-wise Sales')

axes[0,1].plot(df['Month'], df['Sales'],"go--")
axes[0,1].set_xlabel('Month')
axes[0,1].set_ylabel('Sales')
axes[0,1].set_title('Monthly Sales')

axes[1,0].scatter(df['Units_Sold'], df['Sales'],color='green')
axes[1,0].set_xlabel('Units Sold')
axes[1,0].set_ylabel('Sales')
axes[1,0].set_title('Units Sold vs Sales')
axes[1,0].grid('True',alpha=0.5)

axes[1,1].hist(df['Rating'], bins=5,color='green',edgecolor='black')
axes[1,1].set_xlabel('Rating')
axes[1,1].set_ylabel('Frequency')
axes[1,1].set_title('Distribution of Ratings')
plt.tight_layout()
plt.show()
````

````python
import seaborn as sns
````

````python
plt.figure(figsize=(10,5))
sns.barplot(data=df,x = 'Restaurant',y = 'Orders',hue="City", palette="pastel")
plt.title('Number of Orders for Each Restaurant')
plt.show()
````

````python
plt.figure(figsize=(10,5))
sns.countplot(data=df,y='City',hue="City", palette="pastel")
plt.title('Number of Records for Each City')
plt.show()
````

````python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df,x = 'Avg_Bill',y = 'Orders',hue="City", palette="pastel")
plt.title('Relationship Between Avg_Bill and Orders')
plt.show()
````

````python
plt.figure(figsize=(10,5))
sns.boxplot(data=df,y = 'Restaurant',x = 'Avg_Bill',color='green')
plt.title('Avg_Bill of Different Restaurants')
plt.show()
````

````python
plt.figure(figsize=(10,5))
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap='coolwarm')
plt.title('Correlation Between Orders, Avg_Bill, and Rating')
plt.show()
````
